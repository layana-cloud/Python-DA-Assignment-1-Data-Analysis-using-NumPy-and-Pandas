**Numpy Array Operations:**

1. Create a 1D NumPy Array:

* Create a 1D NumPy array named temperatures_w1 for Week 1 with the 
following values: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] 

```python
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1:",temperatures_w1)
```

* import numpy as np- imports the NumPy library.
* np.array([..]) - creats 1D NumPy array named temperatures_w1 for weerk 1.
* print("Week 1:",temperatures_w1)- displays the output of 1D NumPy array.

2. Inspection and Properties: 
* Inspect the shape, data type, and the number of elements in the array. 
Inspect the shape of the temperatures_w1 array.

```python
print("Shape:",temperatures_w1.shape)
print("Data type:",temperatures_w1.dtype)
print("Number of elements:",temperatures_w1.size)
print("Number of dimensions:",temperatures_w1.ndim)
```

* print(temperatures_w1.shape- returns the shape of the array. 7 is the output means 1 row with 7 columns.
* ".dtypes"- returns data type of elements,here floated64.
* ".size"- returns total number of elements.
* ".ndim"- returns number of dimensions,here 1D array so output is 1.

3. Array Operations: 
* Convert the temperatures_w1 from Celsius to Fahrenheit using the 
formula: Fahrenheit = (Celsius * 9/5) + 32 
* Find the maximum, minimum, and mean temperatures for the week.

```python
Fahrenheit=(temperatures_w1*9/5)+32
print("Temperatures in Fahrenheit:",Fahrenheit)
```

```python
print("Temp in Celsius- Max:",temperatures_w1.max(),"Min:",temperatures_w1.min(),"Mean:",temperatures_w1.mean())
print("Temp in Fahrenheit- Max:",Fahrenheit.max(),"Min:",Fahrenheit.min(),"Mean:",Fahrenheit.mean())
```

* Fahrenheit=(temperatures_w1*9/5)+32- converts each Celsius value to Fahrenheit using vectorized operation.
* temperatures_w1.max()/min()/mean() will find the max, min and average temperature in Celsius.
* Fahrenheit.max()/min()/mean() will find the max,min and average temperature in Fahrenheit.

**Pandas Series:**

1. Creating Pandas Series: 
* Create a Pandas Series named marks with the following data: 
* Marks: 95, 92, 89, 85, 80 
* Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd
marks = pd.Series([95,92,89,85,80],index= ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

* import pandas as pd- imports pandas library.
* pd.Series([..],index=[..])- creates a Series named marks with given marks and custom index labels Rank1 to Rank5.
* print(marks)- displays the series.

2. Indexing and Slicing: 
* Use the integer index position to access the mark of the 1st rank student.

```python
print(marks.iloc[0])
```

* marks.iloc[0] - iloc means integer location. It accesses element at position 0 (first element), which is 95 for Rank1.

* Use the loc accessor to retrieve the marks of the top 3 ranks by 
specifying their index labels. 

```python
print(marks.loc[['Rank1','Rank2','Rank3']])
```

* marks.loc[['Rank1','Rank2','Rank3']] - loc means location by label. It retrieves values using index names, here it gets marks for top 3 ranks.

* Use the iloc accessor to retrieve the mark of the 3rd rank student. 

```python
print(marks.iloc[2])
```

* marks.iloc[2] - iloc accesses by integer position. Position 2 means 3rd element (count starts from 0), which is 89 for Rank3.

* Apply a boolean mask to filter the Series and retrieve the ranks where the 
marks are greater than 90. 

```python
print(marks[marks>90])
```

* marks[marks>90] - Filters the Series and keeps only values where mask is True. So it shows Rank1 and Rank2.

3. Manipulating Series: 
* Modify the mark of the student with the 1st rank to 100.

```python
marks['Rank1']=100
print(marks)
```

* marks['Rank1']=100 - Accesses the value with label 'Rank1' and updates it to 100.
* print(marks) - Displays the updated Series.

* Remove the entry corresponding to the last rank from the Series.

```python
marks.pop('Rank5')
print(marks)
```

* marks.pop('Rank5') - pop() removes the entry with label 'Rank5' from the Series and returns its value.
* print(marks) - Displays the Series after removal, now only Rank1 to Rank4 remain.

* Compute the CGPA by dividing each mark in the Series by 10.

```python
cgpa=marks/10
print(cgpa)
```

* cgpa=marks/10 - Divides each value in the marks Series by 10 to calculate CGPA.
* print(cgpa) - Displays the new Series with CGPA value.

**Pandas DataFrame:**

1. Creating Pandas DataFrame: 
* Create a Pandas DataFrame named transactions with the following data:
* TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110 
* ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics' 
* Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 
'South', 'North' 
* Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

```python
transactions=pd.DataFrame({
    "TransactionID":[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory":['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region":['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    "Amount":[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

* transactions=pd.DataFrame({...}) - Creates a DataFrame named transactions from a dictionary. Keys become column names, lists become column values. All 4 columns have 10 values each.
* print(transactions) - Displays the DataFrame in tabular form with auto index 0-9.

2. Data Exploration: 
* Display the transactions DataFrame and its basic information, including 
head, tail, shape, column names, and data types.

```python
print("Head:",transactions.head())
print("Tail:",transactions.tail())
print("Shape:",transactions.shape)
print("Column Names:",transactions.columns.tolist())
print("Data types:",transactions.dtypes)
transactions.info
```

* transactions.head() - Shows first 5 rows by default.
* transactions.tail() - Shows last 5 rows.
* transactions.shape - Returns (rows, columns) -> (10, 4).
* transactions.columns.tolist() - Lists all column names.
* transactions.dtypes - Shows data type of each column (int64, object/string).
* transactions.info()- Shows full summary like non-null count, dtypes, memory.

* Display only the 'ProductCategory' and 'Amount' columns. 

```python
print(transactions[['ProductCategory','Amount']])
```

* transactions[['ProductCategory','Amount']] - Double brackets means selecting multiple columns. It creates a new DataFrame with only those 2 columns.
* print(...) - Displays that filtered DataFrame.

* Retrieve the last 3 columns of the DataFrame using iloc or loc. 

```python
print(transactions.iloc[:,-3:])
```

* transactions.iloc[:, -3:] - iloc = index location, : before comma = take all rows, -3: after comma = take last 3 columns (negative means from end).
* It retrieves all rows with only last 3 columns: ProductCategory, Region, Amount.

* Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

```python
filter_rows = transactions[(transactions['Region'] == 'North') & (transactions['Amount']>200)]
print(filter_rows)
```

* transactions['Region'] == 'North' - Creates True/False mask where Region is North.
* transactions['Amount'] > 200 - Creates True/False mask where Amount > 200.
* & - AND operator, keeps rows where BOTH conditions are True.
* transactions[...] - Filters the DataFrame using that combined mask.
* Result is 3 rows (2, 5, 9) where Region is North AND Amount > 200.

* Find the value counts for the 'ProductCategory' column.

```python
print(transactions['ProductCategory'].value_counts())
```

* transactions['ProductCategory'] - Selects only that one column.
* .value_counts() - Counts how many times each unique value appears.
* Result: Electronics appears 4 times, Clothing 3 times, Furniture 3 times.

* Find the unique values in the 'Region' column.

```python
print(transactions['Region'].unique())
```

* transactions['Region'] - Selects the Region column.
* .unique() - Returns list of distinct values without duplicates.
* Result: 4 unique regions - North, South, East, West.

* Group by 'Region' and find the mean amount for each region.

```python
print(transactions.groupby('Region')['Amount'].mean())
```

* transactions.groupby('Region') - Groups all rows by same Region. So 4 groups: East, North, South, West.
* ['Amount'] - Selects only Amount column from those groups.
* .mean() - Calculates average Amount for each group.
* Result: East average is 375.0, North 287.5, South 250.0, West 190.0

**3. Manipulating the DataFrame:**
* Modify the 'Amount' for the transaction with TransactionID 102 to 165. 

```python
transactions.loc[transactions['TransactionID'] == 102,'Amount'] =165
print(transactions[transactions['TransactionID'] == 102])
```

* transactions['TransactionID'] == 102 finds row where ID is 102.
* .loc[ row_condition, 'Amount'] = locate that row, in the Amount column, =165 overwrites the old value with 165.
* print(transactions[transactions['TransactionID'] == 102]) verifies it changed to 165.

* Add a new column 'Discount' by calculating 10% of the 'Amount'. 

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions.head())
```

* transactions['Amount'] * 0.10 calculates 10% for every row.
* transactions['Discount'] = creates a new column named Discount and stores those values.
* print(transactions.head()) shows first few rows now with the new Discount column.

* Remove the row with TransactionID 109. 

```python
transactions= transactions[transactions['TransactionID'] != 109]
print(transactions)
```

* transactions['TransactionID'] != 109 - Creates True/False for all rows where ID is NOT 109 and keeps only rows where condition is True, so row with 109 is filtered out.

* Delete the 'Discount' column from the DataFrame.

```python
transactions.pop('Discount')
print(transactions)
```

* transactions.pop('Discount') - pop() removes the column named Discount from DataFrame and returns it.
* print(transactions) - Now DataFrame goes back to 4 columns only.
